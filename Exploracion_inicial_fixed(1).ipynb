2. Revisión de campos: analiza los campos de cada fichero, identificando su contenido, tipos de datos y posibles inconsistencias. Evalúa la calidad de los datos, buscando errores comunes como valores nulos, duplicados o inconsistencias en los formatos.
a. No tienes que realizar una revisión profunda y tampoco limpiar los datos en su totalidad, basta con comprobar algunos documentos de muestra.

```python
print('Shape:', ae.shape)
print('Duplicados:', ae.duplicated().sum())

# Tipos de datos — formato tabla vertical
print('\nTipos de datos:')
print(pd.DataFrame({
    'columna': ae.columns,
    'dtype':   ae.dtypes.values,
    'nulos':   ae.isnull().sum().values,
    'nulos_%': (ae.isnull().mean() * 100).round(1).values
}).to_string(index=False))
```

```python
print('Shape:', lic.shape)
print('Duplicados:', lic.duplicated().sum())

# Tipos de datos — formato tabla vertical
print('\nTipos de datos:')
print(pd.DataFrame({
    'columna': lic.columns,
    'dtype':   lic.dtypes.values,
    'nulos':   lic.isnull().sum().values,
    'nulos_%': (lic.isnull().mean() * 100).round(1).values
}).to_string(index=False))
```

```python
print('Shape:', loc.shape)
print('Duplicados:', loc.duplicated().sum())

# Tipos de datos — formato tabla vertical
print('\nTipos de datos:')
print(pd.DataFrame({
    'columna': loc.columns,
    'dtype':   loc.dtypes.values,
    'nulos':   loc.isnull().sum().values,
    'nulos_%': (loc.isnull().mean() * 100).round(1).values
}).to_string(index=False))
```
